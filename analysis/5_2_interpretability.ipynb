```python
from pathlib import Path

base_dir = Path.cwd().parent
print(base_dir)
```

```python
import pandas as pd
pd.set_option('Display.max_colwidth', None)
pd.set_option('Display.max_columns', None)
df = pd.read_csv(f'{base_dir}/meta_csv/5.2_interpretability_and_explainability_meta_results.csv')
```

```python
print('BASELINE ACCURACY:')
print(df['correct'].sum() / df.shape[0])

print('TOP FLIPPED:')
print(df['top_flipped'].sum())
print('TOP FLIPPED AFTER ACCURACY:')
print((df['top_new_pred'] == df['gold']).sum() / df.shape[0])
print('TOP FLIP RATE:')
print(df['top_flipped'].sum() / df.shape[0])

print('RANDOM FLIPPED:')
print(df['rnd_flipped'].sum())
print('RANDOM FLIPPED AFTER ACCURACY:')
print((df['rnd_new_pred'] == df['gold']).sum() / df.shape[0])
print('RANDOM FLIP RATE:')
print(df['rnd_flipped'].sum() / df.shape[0])

print('TOP DROP MEAN')
print(df['top_drop'].mean())

print('RANDOM DROP MEAN')
print(df['rnd_drop'].mean())

print('TOP MINUS RANDOM MEAN')
print((df['top_drop'] - df['rnd_drop']).mean())
```

```python
df[(df['top_new_pred'] != df['gold']) & (df['pred'] == df['gold'])][['top_sentence', 'all_sentences']]
```

```python
pd.set_option('Display.max_colwidth', None)
df[(df['top_new_pred'] != df['gold']) & (df['pred'] != df['gold'])][['top_sentence', 'all_sentences']]
```

```python
df[(df['top_new_pred'] == df['gold']) & (df['pred'] != df['gold'])]
```

```python
pd.set_option('Display.max_colwidth', None)
df_b = pd.read_csv(f'{base_dir}/meta_csv/5.2_interpretability_before_meta_results.csv')
df_a = pd.read_csv(f'{base_dir}/meta_csv/5.2_interpretability_after_meta_results.csv')

df_b[['answer', 'parsed_answer', 'confidence', 'evidence', 'question', 'top_sentence', 'options']].tail(1)
```
